```python
# 1. Import Libraries
import numpy as np
import pandas as pd
```

```python
#dataset
marks = np.array([
    [85, 80, 90],
    [70, 75, 65],
    [92, 88, 95],
    [60, 72, 68],
    [78, 82, 80]
])

marks
```

```python
#Q1. Total marks obtained by each student
total_marks = np.sum(marks, axis=1)
total_marks
```

```python
#Q2. Average marks of each student
average_marks = np.mean(marks, axis=1)
average_marks
```

```python
#Q3. Average marks in each subject
subject_average = np.mean(marks, axis=0)
subject_average
```

```python
#Q4. Highest score in each subject
highest_subject_score = np.max(marks, axis=0)
highest_subject_score
```

```python
#Q5. Lowest score in each subject
lowest_subject_score = np.min(marks, axis=0)
lowest_subject_score
```

```python
#Q6. Students whose average marks are above 80
above_80 = average_marks > 80
above_80
average_marks[average_marks > 80]
```

```python
#Q7. np.where() se Pass/Fail status
status = np.where(average_marks >= 40, "Pass", "Fail")
status
```

```python
#Q8. Highest-performing student ka index
highest_student_index = np.argmax(average_marks)
highest_student_index
```

```python
#Q9. Standard deviation for each subject
subject_std = np.std(marks, axis=0)
subject_std
```

```python
#Q10. Final results ko Pandas DataFrame mein convert karna
df = pd.DataFrame({
    "Python": marks[:, 0],
    "SQL": marks[:, 1],
    "Machine Learning": marks[:, 2],
    "Total Marks": total_marks,
    "Average Marks": average_marks,
    "Status": status
})

df
```
