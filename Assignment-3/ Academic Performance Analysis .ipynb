```python
#Verify Official Semester Totals
calculated_totals = (
    subject_df[subject_df["Include_In_Official_Total"] == True]
    .groupby("Semester")["Total_Marks"]
    .sum()
    .reset_index(name="Calculated_Total_Marks")
)

comparison = pd.merge(
    semester_summary,
    calculated_totals,
    on="Semester"
)

comparison["Difference"] = (
    comparison["Official_Total_Marks"]
    - comparison["Calculated_Total_Marks"]
)

comparison
```

***NumPy Statistical Analysis***

```python
marks = semester_summary["Official_Total_Marks"].to_numpy()

print("Mean Marks:", np.mean(marks))
print("Median Marks:", np.median(marks))
print("Maximum Marks:", np.max(marks))
print("Minimum Marks:", np.min(marks))
print("Standard Deviation:", np.std(marks))
```

```python
#Highest and Lowest Scoring Subjects
highest_subject = subject_df.loc[
    subject_df["Total_Marks"].idxmax()
]

lowest_subject = subject_df.loc[
    subject_df["Total_Marks"].idxmin()
]

print("Highest Scoring Subject:")
print(highest_subject)

print("\nLowest Scoring Subject:")
print(lowest_subject)
```

***Visualizations***

```python
#Semester-wise SGPA Graph
plt.figure(figsize=(8, 5))

plt.plot(
    semester_summary["Semester"],
    semester_summary["SGPA"],
    marker="o"
)

plt.title("Semester-wise SGPA")
plt.xlabel("Semester")
plt.ylabel("SGPA")
plt.xticks(semester_summary["Semester"])
plt.grid(True)

plt.show()
```

```python
#Semester-wise Average — Line Graph
semester_average = (
    subject_df.groupby("Semester")["Total_Marks"]
    .mean()
)

plt.figure(figsize=(8, 5))

plt.plot(
    semester_average.index,
    semester_average.values,
    marker="o"
)

plt.title("Figure 1: Semester-wise Average Marks")
plt.xlabel("Semester")
plt.ylabel("Average Marks")
plt.xticks(semester_average.index)
plt.grid(True)

plt.show()
```

```python
#Subject-wise Average — Bar Graph
subject_average = (
    subject_df.groupby("Subject")["Total_Marks"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(12, 8))

plt.bar(
    subject_average.index,
    subject_average.values
)

plt.title("Figure 2: Subject-wise Average Marks")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")
plt.xticks(rotation=90)

plt.tight_layout()
plt.show()
```

```python
#Semester-wise Total — Bar Graph
plt.figure(figsize=(8, 5))

plt.bar(
    semester_summary["Semester"],
    semester_summary["Official_Total_Marks"]
)

plt.title("Figure 3: Semester-wise Total Marks")
plt.xlabel("Semester")
plt.ylabel("Total Marks")
plt.xticks(semester_summary["Semester"])

plt.show()
```
